```python
counter=0          #除算を行った回数
ptr=0              #得られた素数の数
prime = [None]*500 #素数を格納する配列
prime[ptr] = 2    #最初の素数「2」をセット
ptr += 1
for n in range(3, 1001,2): #対象を奇数に限定
  for i in range(1, ptr): #得られた素数との除算
    counter +=1
    if n % prime[i] ==0:
      break
  else:
    prime[ptr]=n   #得られた素数を登録
    ptr += 1
for i in range(1, ptr): #得られた素数を表示
  print(prime[i])
print(f'除算を行った回数{counter}')
```

```python
counter=0          #除算を行った回数
ptr=0              #得られた素数の数
prime = [None]*500 #素数を格納する配列
prime[ptr] = 2    #最初の素数「2」をセット
ptr += 1
prime[ptr] = 3    #2つ目の素数「3」をセット.
ptr += 1          #以下のループでは奇数のみを扱うため
                  #奇数の素数を初期値として入れておく(偶数2は検査不要となるため)

for n in range(5, 1001,2): #対象を奇数に限定
  i = 1
  while prime[i]*prime[i] <= n:
    counter += 2 #prime[i]*prime[i]とn % prime[i]の2つの計算分
    if n % prime[i] ==0:
      break
    i += 1
  else:
    prime[ptr]=n   #得られた素数を登録
    ptr += 1
    counter += 1 #prime[i]*prime[i]分のみの計算分
for i in range(1, ptr): #得られた素数を表示
  print(prime[i])
print(f'除算を行った回数{counter}')
```
